```python
import pandas as pd
import re
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('punkt') # Download the 'punkt' tokenizer data
nltk.download('punkt_tab') # Download the 'punkt_tab' resource required by word_tokenize

# Initialize stemmer and stopwords
lemmatizer = WordNetLemmatizer() # Retained for potential future use or context, but not used in preprocess_text
stop_words = ["it","The","is","a","It","for","this"]
print("stop words:",stop_words)


def preprocess_text(text):
    # 1. Text Normalization: Convert to lowercase and remove non-alphabetic characters
    text = text.lower() # Convert to lowercase
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers

    # 2. Tokenization
    tokens = word_tokenize(text)

    # 3. Stop Word Removal
    filtered_tokens = [word for word in tokens if word not in stop_words]

    # 4. Custom Stemming: Remove 'ing' suffix only
    # Use a regex to match 'ing' only at the end of a word
    custom_stemmed_tokens = [re.sub(r'ing$', '', word) for word in filtered_tokens]

    return ' '.join(custom_stemmed_tokens)
# Create a sample DataFrame
data = {
    'text': [
        "This is a sample sentence for text preprocessing! It contains numbers like 123 and symbols like #$%",
        "Running and jumping are great exercises. They're good for health.",
        "The quick brown foxes jumped over the lazy dogs.",
        "Python is a popular programming language. It's used for data science."
    ]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("DataFrame after Text Preprocessing (with Custom 'ing' Removal):")
print("\n" + "-"*30 + "\n")

# Apply the preprocessing function to the 'text' column
df['processed_text'] = df['text'].apply(preprocess_text)

print(df)
```
