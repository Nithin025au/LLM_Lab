```python
from transformers import pipeline

# Load pre-trained Transformer classifier
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Sample texts
texts = [
    "I really enjoyed this movie.",
    "The movie was terrible and boring.",
    "The product is amazing.",
    "I did not like the service."
]

# Classify each text
results = classifier(texts)

# Display results
for text, result in zip(texts, results):
    print("Text:", text)
    print("Classification:", result["label"])
    print("Confidence:", result["score"])
    print()
```

```python
# Load pre-trained GPT-2 model
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# Sample prompts
prompts = [
    "Artificial intelligence is changing the world because",
    "The future of space exploration will",
    "Machine learning can help students by"
]

# Generate text
for prompt in prompts:

    result = generator(
        prompt,
        max_new_tokens=50,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7
    )

    print("Prompt:")
    print(prompt)

    print("\nGenerated Output:")
    print(result[0]["generated_text"])

    print("\n" + "-" * 60)
```
