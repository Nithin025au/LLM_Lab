```python
from google import genai
from google.colab import userdata

# Initialize client
client = genai.Client(
    api_key=userdata.get("GOOGLE_API_KEY")
)

# Zero-shot prompt
prompt_zero_shot = """Classify the sentiment of the following movie review:
'This movie was absolutely brilliant! The acting was superb
and the plot kept me on the edge of my seat.'"""

# Generate response
response_zero_shot = client.models.generate_content(
    model="gemini-3.6-flash", # Updated model name
    contents=prompt_zero_shot
)

print("Zero-Shot Response:", response_zero_shot.text)
```

### 2. Few-Shot Prompting

**Few-shot prompting** involves providing the model with a few examples (demonstrations) of the task along with the new input. These examples help the model understand the desired input-output format and the pattern of the task, improving its performance, especially for more complex or nuanced tasks.

The examples act as an in-context learning mechanism, guiding the model towards the desired behavior.

```python
from google import genai
from google.colab import userdata

# Initialize client if not already done (ensure it's only done once)
try:
    client
except NameError:
    client = genai.Client(
        api_key=userdata.get("GOOGLE_API_KEY")
    )

prompt_few_shot = """Classify the following texts into 'Positive', 'Negative', or 'Neutral':

Text: 'I love this product!'
Sentiment: Positive

Text: 'The service was terrible.'
Sentiment: Negative

Text: 'The weather is okay today.'
Sentiment: Neutral

Text: 'What a fantastic experience, I would highly recommend it!'
Sentiment:"""

response_few_shot = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt_few_shot
)
print("Few-Shot Response:", response_few_shot.text)
```

### 3. Instruction-Based Prompting

**Instruction-based prompting** focuses on giving clear, explicit instructions to the model about what it needs to do. This technique is often combined with zero-shot or few-shot approaches, but the emphasis is on the clarity and comprehensiveness of the instructions themselves. You can specify the output format, constraints, and specific actions the model should take.

This is particularly useful when you need precise control over the model's output or when the task is multi-faceted.

```python
from google import genai
from google.colab import userdata

# Initialize client if not already done (ensure it's only done once)
try:
    client
except NameError:
    client = genai.Client(
        api_key=userdata.get("GOOGLE_API_KEY")
    )

prompt_instruction_based = """Summarize the following article in exactly three bullet points. Each bullet point should be a concise sentence.

Article: 'Artificial intelligence (AI) is rapidly transforming various industries. From healthcare to finance, AI-powered solutions are automating tasks, enhancing decision-making, and creating new opportunities. However, concerns about job displacement and ethical implications remain significant. Researchers are actively working on developing responsible AI frameworks and ensuring fair and transparent algorithms.'

Summary:"""

response_instruction_based = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt_instruction_based
)
print("Instruction-Based Response:", response_instruction_based.text)
```
