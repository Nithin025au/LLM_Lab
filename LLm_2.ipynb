```python

import torch
from transformers import AutoTokenizer, AutoModel

model_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
#Sample text
text = "Hello, how are you today? This is a sample sentence."

# Tokenize the text
# return_tensors='pt' ensures PyTorch tensors are returned
# padding=True pads sequences to the longest sequence in the batch
# truncation=True truncates sequences to the maximum input length
inputs = tokenizer(text, return_tensors='pt', padding=True, truncation=True)

# Get the model outputs
with torch.no_grad(): # Disable gradient calculation for inference
    outputs = model(**inputs)

# The last_hidden_state contains the contextualized embeddings for each token
# The shape will be (batch_size, sequence_length, hidden_size)
# For a single sentence, batch_size=1
word_embeddings = outputs.last_hidden_state

print(f"Original text: {text}")
print(f"Tokenized input IDs: {inputs['input_ids']}")
print(f"Shape of word embeddings: {word_embeddings.shape}")

# To get an embedding for the whole sentence (e.g., using the CLS token's embedding)
# The CLS token is typically the first token (index 0) in BERT-like models.
sentence_embedding = word_embeddings[:, 0, :]
print(f"Shape of sentence embedding (CLS token): {sentence_embedding.shape}")
print("\nExample of word embeddings (first token's embedding):")
print(word_embeddings[0, 1, :10].numpy()) # Print first 10 dimensions of the second token's embedding
```
