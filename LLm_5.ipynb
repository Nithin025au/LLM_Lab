```python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    AutoModelForCausalLM
)
import torch

# --------------------------------------------------
# 1. Sample texts
# --------------------------------------------------

texts = [
    "The movie was excellent and I really enjoyed it.",
    "The movie was boring and disappointing.",
    "The product quality is amazing.",
    "The service was terrible."
]


# --------------------------------------------------
# 2. ENCODER-BASED MODEL - BERT
# --------------------------------------------------

print("\n========== ENCODER-BASED MODEL: BERT ==========")

bert_model_name = "bert-base-uncased"

bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)

bert_model = AutoModelForSequenceClassification.from_pretrained(
    bert_model_name,
    num_labels=2
)

# Tokenize input
bert_inputs = bert_tokenizer(
    texts,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

# Prediction
with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs)

bert_predictions = torch.argmax(
    bert_outputs.logits,
    dim=1
)

for text, prediction in zip(texts, bert_predictions):
    label = "Positive" if prediction.item() == 1 else "Negative"
    print(f"Text: {text}")
    print(f"Prediction: {label}\n")


# --------------------------------------------------
# 3. DECODER-BASED MODEL - GPT-2
# --------------------------------------------------

print("\n========== DECODER-BASED MODEL: GPT-2 ==========")

gpt_model_name = "gpt2"

gpt_tokenizer = AutoTokenizer.from_pretrained(gpt_model_name)

# GPT-2 does not have a padding token by default
gpt_tokenizer.pad_token = gpt_tokenizer.eos_token

gpt_model = AutoModelForCausalLM.from_pretrained(
    gpt_model_name
)

# Example prompts
prompts = [
    "The movie was excellent because",
    "The product quality is amazing because",
    "The service was terrible because"
]

for prompt in prompts:

    inputs = gpt_tokenizer(
        prompt,
        return_tensors="pt"
    )

    with torch.no_grad():
        outputs = gpt_model.generate(
            **inputs,
            max_new_tokens=30,
            do_sample=True,
            temperature=0.7,
            pad_token_id=gpt_tokenizer.eos_token_id
        )

    generated_text = gpt_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    print("Prompt:", prompt)
    print("Generated:", generated_text)
    print()
```
